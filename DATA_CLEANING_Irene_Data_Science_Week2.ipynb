```python
# Children — fill with 0 (almost certainly no children)
df['children'] = df['children'].fillna(0)

# country — fill with 'Unknown' (can't infer, but row is still valuable)
df['country'] = df['country'].fillna('Unknown')

# agent — fill with 0 (no agent used)
df['agent'] = df['agent'].fillna(0)

# company — fill with 0 (individual booking, not corporate)
df['company'] = df['company'].fillna(0)
```

**6. STANDARDISE COLUMN NAMES AND DATATYP**ES

**7. HANDLING OUTLIERS**

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["adr"])
plt.title("ADR Before Outlier Removal")
plt.show()
```

```python
Q1 = df["adr"].quantile(0.25)
Q3 = df["adr"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
```

```python
df = df[(df["adr"] >= lower) & (df["adr"] <= upper)]
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["adr"])
plt.title("ADR After Outlier Removal")
plt.show()
```

**DATA CLEANING REPORT**

**Introduction**

The objective of this project was to clean the raw hotel booking dataset and prepare it for further analysis. The dataset contained missing values, duplicate records, inconsistent data types, and outliers. Each issue was addressed using an appropriate data cleaning technique while trying to preserve as much useful information as possible.

-------------------------------------------------------------------------------

**1. Handling Missing Values**

The first step was to check for missing values using isnull() and a heatmap.

Four columns contained missing data:

**children**: Since only four values were missing, I replaced them with 0, assuming that these bookings did not include any children.

**country**: Missing countries were replaced with "Unknown" because there was no reliable way to determine the correct country from other columns.

**agent**: Missing values were replaced with 0, representing bookings that were not made through a travel agent.

**company**: Although this column had a large number of missing values, the missing entries indicate individual bookings rather than missing information. Therefore, I filled these values with 0 instead of removing the column.

After these changes, the dataset no longer contained any missing values.

-------------------------------------------------------------------------------

**2. Removing Duplicate Records**

I checked the dataset for duplicate rows using duplicated().

A total of 31,994 duplicate rows were found. These duplicates were removed using drop_duplicates() to ensure that every booking record appeared only once.

-------------------------------------------------------------------------------

**3. Standardizing the Dataset**

To improve consistency:

All column names were converted to lowercase using snake_case.
The children column was converted from a floating-point value to an integer since the number of children should always be a whole number.

These changes make the dataset easier to read and work with.

-------------------------------------------------------------------------------

**4. Handling Outliers**

Outliers were identified in the adr (Average Daily Rate) column.

A box plot was used to visualize the data, followed by the IQR method to calculate the lower and upper limits.

Records with ADR values outside these limits were removed to reduce the impact of unusually high or low values on future analysis.

-------------------------------------------------------------------------------

**5. Encoding Categorical Variables**

Several categorical columns were converted into numerical format using One-Hot Encoding.

The following columns were encoded:

hotel
meal
customer_type
deposit_type
market_segment

This step prepares the dataset for future machine learning models, which require numerical input instead of text.

-------------------------------------------------------------------------------

**6. Validation**

After completing the cleaning process, I verified the dataset by:

checking for any remaining missing values,
reviewing the updated data types,
examining the summary statistics, and
inspecting the first few rows of the cleaned dataset.

Finally, the cleaned dataset was exported as cleaned_hotel_bookings.csv.

-------------------------------------------------------------------------------

**Conclusion**

After completing the cleaning process, the dataset is more consistent and reliable for analysis. Missing values were handled appropriately, duplicate records were removed, data types were corrected, outliers were treated, and important categorical variables were encoded. The final dataset is now ready for exploratory data analysis or machine learning tasks.
